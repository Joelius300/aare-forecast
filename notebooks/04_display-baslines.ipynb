```python
import pickle
from typing import cast

import matplotlib.pyplot as plt

from aare.evaluation import Forecast
from aare.utils import LAST_PREDICTIONS_FOLDER

plt.rcParams["figure.figsize"] = (16, 9)
```

```python
%load_ext autoreload
%autoreload 2
```

```python
def plot_from_file(path, title: str):
    with open(path, "rb") as file:
        forecast = cast(Forecast, pickle.load(file))
    return forecast.plot(title)
```

```python
def plot_baseline(model: str):
    plot_from_file(LAST_PREDICTIONS_FOLDER / f"{model}.pkl", model)
```

```python
plot_baseline("LOCF")
```

```python
plot_baseline("MEAN")
```

```python
plot_baseline("SNAIVE")
```
