Correspondences:

- tt = smn/tt (air temp in °C)
- ss = smn/ss (sunshine duration in min)
- rr = smn/rr (rain total in mm)
- rh = smn/rh (relative humidity in %)
- dd = smn/dd (wind direction in deg)
- ff = smn/ff (wind speed in km/h)

Pretty simple :)

```python
r = httpx.get("https://aareguru.existenz.ch/rawdata?service=v2018_mdx")
```

```python
r.raise_for_status()
```

```python
body = r.json()
```

```python
bern = body["payload"]["mos"]["BERN"]
bern
```

```python
df = pd.DataFrame.from_dict(bern, orient="index", dtype=np.float32)
df.index = pd.to_datetime(df.index)
df
```

```python
# for later :)
df["wind_y"] = np.sin(df["dd"]) * df["ff"]
df["wind_x"] = np.cos(df["dd"]) * df["ff"]

df
```
