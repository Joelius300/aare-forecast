```python
params = read_params()
```

```python
fs = FeatureSet(
    FEATURES.get_many(
        [
            "temp_bern",
            "temp_bern_diff",
            "temp_bern_sq_neg",
            "temp_bern_neg_sq",
            "temp_bern_diff_abs",
            "temp_bern_diff_sq_neg_neg_sqrt",
        ]
    ),
    split_params=params["split"],
)
fs
```

```python
train = fs.get_train()[0]
len(train)
```

```python
[len(x) for x in train]
```

```python
ts = train[-1]
ts.components
```

```python
df = ts.to_dataframe(time_as_index=False)
df
```

```python
df["temp_bern_diff_manual"] = df["temp_bern"].diff()
```

```python
px.line(df, x=TIME, y=["temp_bern_diff", "temp_bern_diff_manual"])
```

```python
# one is not equal because diff is done before splitting and if you do it after you'll get a NaN
(df["temp_bern_diff"] == df["temp_bern_diff_manual"]).describe()
```

```python
(df["temp_bern_diff"].abs() == df["temp_bern_diff_abs"]).describe()
```

```python
(df["temp_bern_neg_sq"] == -df["temp_bern_sq_neg"]).describe()
```
